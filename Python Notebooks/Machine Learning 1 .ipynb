```python
import pandas as pd
melburne_data = pd.read_csv("https://raw.githubusercontent.com/dphi-official/First_ML_Model/master/melbourne_data.csv")
melburne_data.columns
```

```python
melburne_data.head()
```

```python
y = melburne_data.Price
y.head()
```

We will choose Features which are used to determine the price of the house

```python
melburne_features = ['Rooms', 'Bathroom', 'Landsize', 'Lattitude', 'Longtitude']
X = melburne_data[melburne_features]
X.head()
```

```python
X.describe()
```

Building our First Machine Learning Model

```python
from sklearn.tree import DecisionTreeRegressor
model = DecisionTreeRegressor(random_state = 1)
model.fit(X,y)
```

In the above step the model has now been trained now lets test it by predicting the prices using the Feature variables

```python
model.predict(X)
```
