```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv('https://raw.githubusercontent.com/aiplanethub/Datasets/refs/heads/master/kc_house_data/kc_house_data.csv')
data.head()
```

```python
data.info()
```

```python
data.isnull().sum()
```

```python
from sklearn.model_selection import train_test_split
x = data['price']
y = data.drop('price',axis =  1)
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.3,random_state = 43)
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(y_train,x_train)
```

```python
from sklearn.metrics import r2_score
y_pred = lr.predict(y_test)
print("The r2 score for our model was :",r2_score(x_test,y_pred))
```

Decision Tree Regressor

```python
from sklearn.tree import DecisionTreeRegressor
dtr = DecisionTreeRegressor(random_state = 42)
dtr.fit(y_train,x_train)
```

```python
from sklearn.metrics import r2_score
y_pred = dtr.predict(y_test)
print("The r2 score for our model was :",r2_score(x_test,y_pred))
```
