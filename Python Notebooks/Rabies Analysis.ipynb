Variables that affect dog rabies vaccination

1. Breed vs Vaccination Status

```python
data['Type of Breed'].value_counts()
```

We can see that the Local breed had 42 breeds of the dogs.

```python
data['How many of your dogs are vaccinated against rabies?'].groupby(data['Type of Breed']).sum()
```

The mixed breed was the most vaccinated breed as at 58 compared to the Pure breed.

2. Education level

```python
data['How many of your dogs are vaccinated against rabies?'].groupby(data['Highest Level of Education']).sum()/data['How many of your dogs are vaccinated against rabies?'].sum()*100
```

74.167% of owners who had achieved university education vaccinated their dogs while the ones whose higherst level of education was secondary only 8.8% vaccinated their dogs. This would mean that level of education is an important factor in vaccination.

3. Dog's age

```python
data['How many of your dogs are vaccinated against rabies?'].groupby(data['Age Group']).sum()/data['How many of your dogs are vaccinated against rabies?'].sum()*100
```

From the analyis the youthful age bracket (18-35) have a higher percentage of vaccinating their dogs compared to the other age groups . Hence age of the owner is actually an important factor for vaccination

```python
data['Have you heard about rabies before?'].value_counts()
```

Most of the respodents have a knowledge of Rabies while only 4 don't have a knowledge of Rabies. But then lets see how does this affect vaccination status of their dog. We will now use a crosstab for this.

```python
data['How many of your dogs are vaccinated against rabies?'].groupby(data['Have you heard about rabies before?']).sum()/data['How many dogs do you currently own?'].sum()*100
```

```python
pd.crosstab(data['Has your dog ever been vaccinated?'],data['Have you heard about rabies before?'])
```

From the analysis we had 71 people who knew about rabies and did vaccinate their dogs. Meanwhile we had 26 people who knew about rabies but did not vaccinate their dogs. This propotion would explain to us that also the knowledge about rabies is an important factor
