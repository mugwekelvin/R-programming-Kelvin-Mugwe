```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
data = pd.read_csv('https://raw.githubusercontent.com/dphi-official/Datasets/master/framingham.csv')
data.head()
```

```python
pd.crosstab(data.age,data.education)
```

```python
# We will drop Education asit does not affect heart disease
data = data.drop('education',axis = 1)
data
```

```python
data.info()
```

```python
data = data.rename(columns = {'male':'Gender_Male'})
data.info()
```

```python
# Plotting the graph of the people with heart disease
sns.countplot(x = data['TenYearCHD'])
plt.grid(True)
plt.title('People with heart disease')
plt.show()
```

```python
data = data.dropna()
```

```python
import sklearn
new_features=data[['age','Gender_Male','cigsPerDay','totChol','sysBP','glucose','TenYearCHD']]
x = new_features.iloc[:,:-1]
y = new_features.iloc[:,-1]
```

```python
print(x.shape)
print(y.shape)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.3,random_state = 5)
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
model.fit(x_train,y_train)
```

```python
# let's predict using the test data
model.predict(x_test)
```

```python
# Checking for accuracy of our model
from sklearn.metrics import accuracy_score
accuracy_score(y_test,model.predict(x_test))
```

```python
# We will now use the Confusion Matrix to check for false positives ,false negatives,true positives and true negatives
from sklearn.metrics import confusion_matrix
confusion_matrix(y_test,model.predict(x_test))
```

```python
# Presenting the confusion matrix in a heatmap
cm = confusion_matrix(y_test,model.predict(x_test))
conf_matrix = pd.DataFrame(data =cm , columns = ['Predicted : 0','Predicted :1'],index = ['Actual : 0','Actual : 1'])
sns.heatmap(conf_matrix ,annot = True)
plt.title('Confusion Matrix')
plt.show()
```
