# NumPy Tutorial
This notebook provides an introduction to NumPy, its installation, core concepts like arrays, and various operations including mathematical, statistical, and matrix computations.

## What is NumPy?
NumPy (Numerical Python) is a powerful library for numerical computations in Python. It provides support for arrays, matrices, and many mathematical functions.

## Uses of NumPy
- Handling large multidimensional arrays and matrices
- Mathematical and logical operations
- Fourier transforms and routines for shape manipulation
- Operations related to linear algebra
- Integration with other libraries like Pandas, Matplotlib, Scikit-learn

````
import numpy as np
print("NumPy version:", np.__version__)
````

## Creating NumPy Arrays

````
# 1D Array
arr1d = np.array([1, 2, 3, 4])
print("1D Array:", arr1d)
````

````
# 2D Array
arr2d = np.array([[1, 2], [3, 4]])
print("2D Array:")
print(arr2d)
````

````
# 3D Array
arr3d = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
print("3D Array:")
print(arr3d)
````

## Mathematical Operations on NumPy Arrays

````
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print("Addition:", a + b)
print("Multiplication:", a * b)
print("Exponentiation:", a ** 2)
````

## Statistical Operations on NumPy Arrays

````
arr = np.array([[1, 2, 3], [4, 5, 6]])
print("Mean:", np.mean(arr))
print("Median:", np.median(arr))
print("Standard Deviation:", np.std(arr))
````

## Matrix Operations with NumPy

````
mat1 = np.array([[1, 2], [3, 4]])
mat2 = np.array([[5, 6], [7, 8]])
print("Matrix Multiplication:")
print(np.dot(mat1, mat2))
````

## Plotting NumPy Arrays

````
import matplotlib.pyplot as plt
x = np.linspace(0, 10, 100)
y = np.sin(x)
plt.plot(x, y)
plt.title("Sine Wave")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.grid(True)
plt.show()
````

## Flattening and More Operations on NumPy Arrays

````
arr = np.array([[1, 2], [3, 4]])
print("Original Array:")
print(arr)
print("Flattened Array:", arr.flatten())
print("Transpose:")
print(arr.T)
````
