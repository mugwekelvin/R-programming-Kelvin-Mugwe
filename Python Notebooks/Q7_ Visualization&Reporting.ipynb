# We may be interested for the year 2023 over the months for this I will use Plotly

```python
import plotly.express as px
```

```python
from datetime import datetime
def convert_date(x):
  return datetime.strptime(x, '%m/%d')
```

```python
data = pd.read_csv('Data.csv')
```

```python
data['Transaction Date'] = pd.to_datetime(data['Transaction Date'])
```

```python
data_2023 = data[data['Transaction Date'].dt.year == 2023].copy()
display(data_2023.head())
```

```python
fig = px.line(data_2023, x='Transaction Date', y='Transaction Amount', title='Transaction Amount Over Time')
fig.update_xaxes(rangeslider_visible=True)
fig.show()
```

Since the plot is not understandable we need to agreggate per month for the year 2023

```python
monthly_data_2023 = data_2023.groupby(pd.Grouper(key='Transaction Date', freq='M'))['Transaction Amount'].sum().reset_index()

fig = px.line(monthly_data_2023, x='Transaction Date', y='Transaction Amount', title='Total Transaction Amount by Month in 2023')
fig.update_xaxes(title='Month',rangeslider_visible = True)
fig.update_yaxes(title='Total Transaction Amount')
fig.show()
```

This plot shows that for a particular date where the Transaction Amount is high

# Bar Graph

Agreggating the Transaction Amount vs Account Type

```python
data_2023 = data[data['Transaction Date'].dt.year == 2023].copy()
data_filtered = data_2023.groupby('Account Type')['Transaction Amount'].sum().reset_index()
```

```python
fig = px.bar(data_filtered, x='Account Type', y='Transaction Amount', title='Account Type vs Transaction Amount 2023')
fig.update_xaxes(title='Account Type')
fig.update_yaxes(title='Transaction Amount')
fig.show()
```

The Current Account Type had the highest Transaction Account(6299351.17) compared to the Savings Account(6243154.55)

# Bar Graph

which Loan Type has the largest interest rate

```python
data_loan_int = data.groupby('Loan Type')['Interest Rate'].mean().reset_index()
```

```python
fig = px.bar(data_loan_int, x='Loan Type', y='Interest Rate', title='Loan Type vs Interest Rate')
fig.update_xaxes(title='Loan Type')
fig.update_yaxes(title='Interest Rate')
fig.show()
```

The Personal Loan Type had the highest Average Interest Rate at 5.657% compared to the Auto at 5.45 %

# Bar Graph

I would be interested in which are the poorly performing branches according to the Transaction Amounts

```python
data['Branch ID'] = data['Branch ID'].astype('object')
data_br_trans = data.groupby('Branch ID')['Transaction Amount'].sum().reset_index()
```

```python
data_bar_trans_desc = data_br_trans.sort_values(by='Transaction Amount', ascending=False).tail(10)
```

```python
fig = px.bar(data_bar_trans_desc, x='Branch ID', y='Transaction Amount', title='Branch ID vs Transaction Amount')
fig.update_xaxes(title='Branch ID')
fig.update_yaxes(title='Transaction Amount')
fig.show()
```

Branch 40 had the lowest Transaction Amount over the two years as at 78.155K hence proper marketing should be done at the Branch to enable it.
