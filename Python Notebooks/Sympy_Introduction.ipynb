### Double Integration

Consider a bivariate probability density function (PDF) defined as follows:

$f(x, y) = \begin{cases}
k(x + y) & \text{if } 0 \le x \le 1 \text{ and } 0 \le y \le 1 \\
0 & \text{otherwise}
\end{cases}$

Find the value of $k$

$\displaystyle \int_{0}^{1}\int_{0}^{1} k  \cdot (x+y) = 1 $

```python
# since total probability on the given interval is 1 then:
from sympy import *
x,y,k = symbols('x y k')
f = k*(x+y)
equation = integrate(f,(x,0,1),(y,0,1))-1
```

```python
k_value = solve(equation , k)[0]
k_value
```

Find the $E[X]$

```python
E_x = integrate(x*f*k_value,(x,0,1),(y,0,1))
E_x = E_x.subs({k:k_value})
E_x
```

Find the $E[x^{2}]$

```python
Ex_2 = integrate(x**2*f*k_value,(x,0,1),(y,0,1))
Ex_2 = Ex_2.subs({k:k_value})
Ex_2
```

Find $Var[X]$

```python
Var_x = simplify(Ex_2 -  (E_x**2))
Var_x
```

## Using Sympy to derive Probability Distribution Functions

### Discrete Probability Distributions

1 . Poisson Distribution
Derive the mean , variance of the poisson distribution given the pdf is
$\dfrac{\lambda^{k} \cdot e^{-\lambda}}{k!}$

```python
from sympy import *

Lambda = Symbol('Lambda', positive=True)
k = Symbol('k', positive=True, integer=True)
pdf = (Lambda**k * exp(-Lambda))/factorial(k)
mean = summation(k*pdf,(k,0,oo))

mean
```

```python
E_x_2 = summation(k**2*pdf,(k,0,oo))
E_x_2
```

```python
simplify(E_x_2 - mean**2)
```

2. Bernoulli's Distribution where PDF is
$P(X=x) = p^{x} \times (1-p)^{(1-x)}$

With the PDF find the mean and the variance

```python
p = Symbol('p')
x = Symbol('x')
pdf = (p**x)*(1-p)**(1-x)
mean = summation(x*pdf,(x,0,1))
mean
```

```python
Ex_2 = summation(x**2 * pdf,  (x,0,1))
simplify(Ex_2 - (mean)**2)
```

## Continous Probability Distributions

### 1.  Uniform  Population Distribution

$\displaystyle  f(x) = \dfrac{1}{b-a}$

```python
b = Symbol('b')
a = Symbol('a')
x = Symbol('x')
pdf = 1/(b-a)
mean = integrate(x*pdf,(x,a,b))
mean = simplify(mean)
mean
```

```python
Ex_2 = integrate(x**2  * pdf ,  (x,a,b))
Ex_2
```

```python
Ex_2 = simplify(Ex_2)
Ex_2
```

```python


# Calculate variance and simplify
variance = simplify(Ex_2 - mean**2)
variance = factor(expand(variance))  # Expand and then factor

# Display variance
variance
```

Hence the Dsitribution is as follows $X \sim U\left(\frac{a+b}{2},\frac{(a-b)^{2}}{12}\right)$

2 . Normal Probability Function


Find the mean and variance given the PDF as
$f(x) = \dfrac{1}{\sigma \sqrt{2\pi}} e^{\dfrac{-(x-\mu)}{2\sigma^{2}}}$


```python
sigma = Symbol('sigma',positive = True)
mu =Symbol('mu',positive = True)
pi = Symbol('pi',positive = True)
x = Symbol('x')
pdf = 1/(sigma*sqrt(2*pi))*exp(-(x-mu)**2/(2*sigma**2))
mean = integrate(x*pdf,(x,-oo,oo))
mean
```

when simplified it becomes $\mu $

```python
Ex_2 = integrate(x**2 * pdf, (x,-oo,oo))
simplify(Ex_2)
```

```python
cancel(factor(Ex_2 - mean**2))
```

When simplified it becomes $\sigma ^2$

Hence the Distribution is as follows $ X \sim N(\mu,\sigma^2)$
