# 5. Data Imputation (Handling Missing Values)

Fill Missing Values with Mean (Penguins)

```python
penguins['bill_length_mm'] = penguins["bill_length_mm"].fillna(penguins["bill_length_mm"].mean())
```

Fill Missing Values with Median (Iris)

```python
penguins['bill_length_mm'].fillna(penguins['bill_length_mm'].median(), inplace=True)
```

# Backward Fill

```python
penguins['bill_depth_mm'].fillna(penguins['bill_depth_mm'].bfill(), inplace=True)
```

# Forward Fill

```python
penguins['bill_depth_mm'].fillna(penguins['bill_depth_mm'].ffill(), inplace=True)
```

Fill Missing Values with Mode (Categorical Data)

```python
penguins["sex"].fillna(penguins["sex"].mode()[0], inplace=True)
```

## Challenge (iris)

Rename species to flower_type.

```python
# put your code here
iris.rename(columns={"species":"flower_type"},inplace = True)
```

```python
iris.info()
```

Create a new column size_category:

"Small" if sepal_length < 5,

"Medium" if between 5 and 6.5,

"Large" otherwise.

```python
def size_category_func(row):
  if row['sepal_length'] < 5:
    return 'Small'
  elif row['sepal_length'] >= 5 and row['sepal_length'] <= 6.5:
    return 'Medium'
  else:
    return 'Large'

iris['size_category'] = iris.apply(size_category_func, axis=1)
iris.head(5)
```

# 14. Data Encoding Using map()

Encoding Sex in Penguins Dataset (Male: 1, Female: 0)

```python
penguins['sex'] = penguins["sex"].fillna(penguins['sex'].mode()[0])
```

```python
penguins["sex_encoded"] = penguins["sex"].fillna(penguins['sex'].mode()[0]).map({"MALE": 1, "FEMALE": 0})
penguins.head().dropna()
```

Encoding Species in Iris Dataset

```python
iris["species_encoded"] = iris["flower_type"].map({"setosa": 0, "versicolor": 1, "virginica": 2})
iris.head()
```

# Removing Duplicates

```python
import pandas as pd

data = {
    "ID": [1, 2, 2, 3, 4, 4, 5, 6, 6],
    "Name": ["Alice", "Bob", "Bob", "Charlie", "David", "David", "Eve", "Frank", "Frank"],
    "Age": [24, 30, 30, 35, 40, 40, 29, 33, 33],
    "City": ["Nairobi", "Mombasa", "Mombasa", "Kisumu", "Nakuru", "Nakuru", "Eldoret", "Thika", "Thika"]
}

df = pd.DataFrame(data)
print(df)
```

```python
df['Age'].mean()
```

```python
df[df.duplicated()]
```

```python
df.drop_duplicates(inplace=True)
```

```python
df = df.reset_index()
df
```
