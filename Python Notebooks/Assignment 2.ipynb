```python
import pandas as pd
import numpy as np
import sklearn
```

```python
# Importing and previewing data
data = pd.read_csv('https://archive.org/download/retail_dataset/retail_dataset.csv')
data.head()
```

```python
#Displaying the column data types
data.info()
```

```python
# Displaying the columns with duplicates
data[data.duplicated()]
```

```python
# Now let's remove the duplicated data
data.drop_duplicates(inplace = True)
data.head()
```

```python
#Displaying shape of our data after dropping duplicates
data.shape
```

Handling Missing Values

```python
# Displaying the Distribution of missing values
data.isnull().sum()
```

```python
# Displaying total missing values in our dataset
data.isnull().sum().sum()
```

Handling missing data

```python
# Removing missing data
rm_data = data.dropna().isnull().sum()
```

```python
# Mean and median of data before imputation
print("The mean price was",np.mean(data['Price']))
print("The median quantity is ",np.median(data['Quantity']))
```

There are no missing values after removal of all missing values

```python
# Handling missing data by imputation of missing data by the mean
data['Price'] = data['Price'].fillna(data['Price'].mean())
```

```python
data['Quantity'] = data['Quantity'].fillna(data['Quantity'].median())
```

```python
data.isnull().sum()
```

```python
data.iloc[:,3:5]
```

There are no missing values in our data frame

Data Scaling and Normalization

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
new_transform = scaler.fit_transform(data.iloc[:,3:5])
new_transform = pd.DataFrame(new_transform,columns = ['Price','Quantity'])
new_transform.head()
```

The data has now been transformed and is scaled between 0 and 1

```python
# Let's see the value count for the Region column to be able to be Coded
data['Region'].value_counts()
```

Recoding the Regions ready for ML

```python
categorical_mapping = {'North':1,'West':2,'South':4,'East':3}
data['Region'] = data['Region'].map(categorical_mapping)
data.head()
```
