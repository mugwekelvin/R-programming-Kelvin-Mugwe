```python
import sympy
from sympy import *
x = Symbol('x')
y = Symbol('y')
eq_1 = 2*x + y - 100
eq_2 = x + 2*y -100
solve([eq_1,eq_2], x,y)
```

```python
z = Symbol('z')
eq_1 = x + y + z -1
eq_2 = 2*x + y - 1
eq_3 = 5*x + 3*y + 2*z -4
solve([eq_1,eq_2,eq_3],x,y,z)
```
