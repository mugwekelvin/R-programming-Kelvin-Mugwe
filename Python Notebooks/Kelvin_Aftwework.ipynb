Create a program that performs some computation using the python programming language. Tigo Telcom provides a mobile subscription plan with 200 minutes of airtime and 200 text messages for  10amonth.Each
 additional minute of airtime costs .03, and other text messages cost $.001 each. The customer's bill is also subject to 14% VAT Tax. Write a python program that will read the no. of minutes and text messages used in a month, then displays the base charge, additional minutes (if any), other text messages (if any), the tax charged, and the total bill. A few things to note:

Your input should be in the form of a dictionary.
Your output should be more like an SMS message.
Display the charges in 2 decimal places.

```python
no_of_minutes = eval(input("Enter the no of minutes you want to use "))
no_of_messages = eval(input("Enter the no of messages you want to send "))


if (no_of_minutes > 200):
  additional_minutes = no_of_minutes - 200
  extra_minute_cost = 0.03*additional_minutes
else :
  additional_minutes = 0
  extra_minute_cost = 0

if(no_of_messages > 200):
  other_messages = no_of_messages - 200
  extra_text_cost = 0.001*other_messages
else:
  other_messages = 0
  extra_text_cost = 0

# Inout storage
subscription_plan['airtime']= no_of_minutes
subscription_plan['text_messages'] = no_of_messages
subscription_plan = {
    'airtime' :  no_of_minutes,
    'text_messages': no_of_messages
}
# Declaration of Variables
base_charge = 10.00
vat_rate = 0.14
total_cost_b4_VAT = base_charge + extra_minute_cost + extra_text_cost
VAT_payable =round(total_cost_b4_VAT *vat_rate,2)
total_cost_after_VAT =  total_cost_b4_VAT + VAT_payable
# Prints the result like a Receipt
def printed():
  print("-------------MONTHLY BILL BREAKDOWN --------------")
  print(f"Airtime used : {subscription_plan['airtime']} minutes")
  print('---------------------------------------------------')
  print(f"Text messages sent : {subscription_plan['text_messages']} messages ")
  print("------------SMS BILL-------------------- ")
  print(f"Base Charges :${base_charge}")
  print(f"Additioanal Airtime per minute  COST : ${extra_minute_cost} ")
  print(f"Additional Message Cost :${extra_text_cost}")
  print(f"VAT payable : ${VAT_payable}")
  print(f"Total Cost : ${round(total_cost_after_VAT,2)}")
  print("------------THANK YOU FOR CHOOSING TELECOM------------------------------")
# Invoking the function
printed()





```
