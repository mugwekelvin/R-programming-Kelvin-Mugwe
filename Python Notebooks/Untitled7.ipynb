```python
# prompt: Derive the poisson distribution using Sympy

import sympy
from sympy import Symbol, factorial, exp

# Define the symbols
n = Symbol('n', integer=True, positive=True)
k = Symbol('k', integer=True, nonnegative=True)
lam = Symbol('lambda', positive=True)

# Define the Poisson distribution
poisson_pmf = (lam**k / factorial(k)) * exp(-lam)

# Display the distribution
print(f"The Poisson Distribution is: P(X = k) = {poisson_pmf}")
```

```python
from scipy.stats import poisson

# Define the Poisson distribution parameter (lambda)
# Example value
λ=10
# Calculate mean and variance
mean = poisson.mean(λ)
variance = poisson.var(λ)

print(f"Mean: {mean}")
print(f"Variance: {variance}")
```

```python
from scipy.stats import bernoulli

# Define probability of success
p = 0.6

# Create Bernoulli random variable
rv = bernoulli(p)

# Calculate mean and variance using SciPy
mean = rv.mean()
variance = rv.var()

print(f"Mean (SciPy): {mean}")
print(f"Variance (SciPy): {variance}")

# Calculate mean and variance using derived formulas
derived_mean = p
derived_variance = p * (1 - p)

print(f"Mean (Derived): {derived_mean}")
print(f"Variance (Derived): {derived_variance}")
```
