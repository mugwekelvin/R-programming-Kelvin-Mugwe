```python
import seaborn as sns
data = sns.load_dataset('iris')
data.head()
```

```python
data.info()
```

```python
data.describe()
```

```python
import matplotlib.pyplot as plt
datas = data.drop('species',axis = 1)
sns.heatmap(datas.corr(),annot = True)
plt.show()
```

```python
# plotting a linear regression model for petal_width and petal_length
sns.regplot(x = datas['petal_length'],y=datas['petal_width'])
plt.show()
```

```python
# countplot for species
sns.countplot(x=data['species'])
plt.show()
```

```python
# Pie chart
# Pie chart
# Calculate the count of each species
species_counts = data['species'].value_counts()

# Create the pie chart using the counts
plt.pie(species_counts, labels=species_counts.index)
plt.legend()
plt.show()
```
