# 1. Scatter Plot

This plot displays the relationship between two variables, sepal_length and sepal_width. Each point represents a single flower, and its position on the plot is determined by its measurements. It's an effective way to see if there is a correlation or a pattern between the two variables.

A bar plot represents categorical data with rectangular bars, where the height of each bar is proportional to the value it represents.

# 3. Countplot

This is a specialized bar plot from Seaborn. It shows the counts of observations in each categorical bin. Essentially, it's a histogram for categorical data. It is a simple and effective way to see the distribution of categorical variables and is particularly useful for quickly checking if a dataset is balanced across different categories. For example, a count plot of the Iris species would show you exactly how many flowers of each species are in the dataset.



```python
sns.countplot(data = penguins , x = 'species',hue = 'island')
plt.xlabel('Species')
plt.ylabel('Count')
plt.title('Count of Penguins by Species and Island')
plt.show()
```

6. Boxplot

 Also known as a box-and-whisker plot, this visualization shows the distribution of a variable through its five-number summary: minimum, first quartile (Q1), median, third quartile (Q3), and maximum. It is particularly useful for identifying potential outliers and for comparing the distributions of multiple groups.

```python
import numpy as np
q1 = np.quantile(iris['sepal_length'],0.25)
q3 = np.quantile(iris['sepal_length'],0.75)
iqr = q3-q1
max_value = np.max(iris['sepal_length'])
min_value = np.min(iris['sepal_length'])
median = np.median(iris['sepal_length'])
upper_whisker = max_value + 1.5 *iqr
lower_whisker = min_value - 1.5*iqr
print(upper_whisker)
print(lower_whisker)
print(median)
```

```python
plt.boxplot(iris['sepal_length'])
plt.ylabel('Sepal Length (cm)')
plt.title('Boxplot of Sepal Lengths')
plt.show()
```

# Joint plot

A joint plot is a powerful Seaborn visualization that displays the relationship between two variables and their individual distributions. It's a combination of a scatter plot (or other bivariate plot) in the center and histograms (or KDE plots) on the margins. This makes it an excellent tool for simultaneously exploring both the bivariate and univariate distributions of a dataset.

```python
iris.info()
```

```python
sns.jointplot(data = iris, x= 'sepal_length',y = 'petal_length',kind = 'kde',color = 'maroon')
plt.show()
```

```python
sns.jointplot(data = penguins , x = 'flipper_length_mm',y = 'body_mass_g',kind = 'reg')
plt.show()
```

# Regression Plot

A regression plot, or regplot in Seaborn, is a visualization that displays the relationship between two numerical variables while fitting and showing a linear regression model. It’s a powerful tool for quickly assessing the strength and direction of a linear relationship between variables and for visualizing the line of best fit. .



```python
sns.regplot(data = iris, x = 'sepal_length',y = 'sepal_width')
plt.title("Regression plot of Seoal Length vs Sepal Width")
plt.show()
```

We can now get the intercept and gradient from the plot

```python
from scipy.stats import linregress

# Perform linear regression
slope, intercept, r_value, p_value, std_err = linregress(iris['sepal_length'], iris['sepal_width'])

print(f"Intercept: {intercept}")
print(f"Gradient (Slope): {slope}")
```

# Pie Chart

A pie chart is a circular statistical graphic divided into slices to illustrate numerical proportion. In a pie chart, the arc length of each slice is proportional to the quantity it represents. They are best used to show the composition of a whole at a single point in time, especially with a small number of categories.

```python
species_count_df = penguins['species'].value_counts().reset_index()
species_count_df.columns = ['species','count']
species_count_df
```

```python
plt.pie(species_count_df['count'],labels = species_count_df['species'],autopct = '%1.1f%%')
plt.show()
```
