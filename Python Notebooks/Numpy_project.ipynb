# Regression Modelling

$$\beta = (X^{T}\cdot X)^{-1}(x^{T}y)$$

```python
X = np.array(data.drop('y',axis=1))
y = np.array(data['y'])
```

```python
X
```

```python
X_transpose = X.T
A = np.dot(X_transpose,X)
```

```python
B = np.dot(X_transpose,y)
```

```python
Beta_Coefficients = np.dot(np.linalg.inv(A),B)
Beta_Coefficients
```

$$
\beta =
\begin{bmatrix}
\beta_0 \\
\beta_1 \\
\beta_2 \\
\beta_3
\end{bmatrix}
=
\begin{bmatrix}
\text{intercept} \\
2.62755 \\
-1.05824 \\
1.00150
\end{bmatrix}
$$

# Finding the Intercept


 $$\beta_{0} = \bar{y} - \sum_{i=1}^{3}\beta_{i}\cdot \bar{X_{i}}$$

```python
mean_x = np.mean(X)
mean_x

```

```python
X
```

```python
# Calculate the mean of each feature in X
mean_X = np.mean(X, axis=0)

# Calculate the intercept (Beta_0)
Beta_0 = np.mean(y) - np.sum(Beta_Coefficients * mean_X)

print(f'Intercept (Beta_0): {Beta_0}')
```

Now that you have the intercept ($\beta_0$) and the coefficients ($\beta_1, \beta_2, \beta_3$), you can write the regression equation:

$$ y = \beta_0 + \beta_1 X_1 + \beta_2 X_2 + \beta_3 X_3 $$

Substituting the calculated values:

```python
print(f'Regression Equation: y = {Beta_0:.4f} + {Beta_Coefficients[0]:.4f}X1 + {Beta_Coefficients[1]:.4f}X2 + {Beta_Coefficients[2]:.4f}X3')
```
