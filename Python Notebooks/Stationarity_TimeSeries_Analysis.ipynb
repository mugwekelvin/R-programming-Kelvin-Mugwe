```python
import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np
import os
from datetime import datetime
import pandas as pd
import plotly.express as px

import warnings
warnings.filterwarnings('ignore')


mpl.rcParams['figure.figsize'] = (10, 8)
mpl.rcParams['axes.grid'] = False
```

```python
def date_parse(x):
  return datetime.strptime(x,"%m/%d/%Y")
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/srivatsan88/YouTubeLI/master/dataset/amazon_revenue_profit.csv',parse_dates=['Quarter'],date_parser = date_parse)
df.head()
```

```python
fig = px.line(df,x='Quarter',y = 'Revenue',title = "Quarterly Revenue Returns")
fig.update_xaxes(rangeslider_visible = True)
fig.show()
```

From the graph this displays that the time series is a non-stationary type though we can prove it by Testing for Stationarity using KPSS and ADFULLER

## 1. KPSS TEST

$H_{0} : \textbf{The series is stationary} \qquad v/s \qquad H_{1} : \textbf{The series is not stationary}  $

```python
from statsmodels.tsa.stattools import kpss
stats,p,lags,critical_values = kpss(df['Revenue'],'ct')
```

```python
print(f'Test Statistics : {stats}')
print(f"p_value: {p}")
print("Critical value : {critical_values}")
if p<0.05:
  {
      print("Series is not stationary \n")
  }
else:
 {
      print("Series is stationary ")
  }
```

## 2.ADFULLER TEST

$H_{0} : \textbf{Series possesses a unit root and hence is not stationary} \quad v/s \quad H_{a} :\textbf{Series is stationary}$

```python
from statsmodels.tsa.stattools import adfuller
result = adfuller(df['Revenue'])
```

```python
print(result)
```

```python
print(f'Test Statistic : {result[0]}')
print(f'p-value : {result[1]}')
print(f'Critical Values : {result[4]}')
if result[1]>0.05:
  print("Series is not stationary \n")
else:
  print("Series is stationary ")
```
