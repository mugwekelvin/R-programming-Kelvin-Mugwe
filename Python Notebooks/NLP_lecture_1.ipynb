Stemming and Lemmatization

Stemming reduces words to their root form by removing suffixes while lemmatization brings words to their base or dictionary form.

```python
from nltk.stem import PorterStemmer
pst = PorterStemmer()
```

```python
pst.stem('having')
```

```python
pst.stem('affection')
```

```python
pst.stem('denied')
```

```python
pst.stem('playing')
```

```python
pst.stem('giving')
```

```python
from nltk.stem import LancasterStemmer
from nltk.stem import LancasterStemmer
lst = LancasterStemmer()
words = ['give','giving','given','gave']
for i in range(len(words)):
  print(words[i] + ":",lst.stem(words[i]))
  i = i+1
```

```python
from nltk.stem import SnowballStemmer
sbst = SnowballStemmer('english')

for i in range (len(words)):
  print(words[i] + ":",sbst.stem(words[i]))
  i = i+1
```

```python
from happytransformer import HappyTextToText, TTSettings
```

```python
happy_tt = HappyTextToText("T5", "vennify/t5-base-grammar-correction")
```

```python
happy_tt = HappyTextToText("T5", "vennify/t5-base-grammar-correction")
```

```python
args = TTSettings(num_beams=5, min_length=1, max_length=100)
text = "The cat is sit on the mat."  # Example sentence with a grammatical error

result = happy_tt.generate_text(text, args=args)

corrected_sentence = result.text

print(f"Original sentence: {text}")
print(f"Corrected sentence: {corrected_sentence}")
```
