There are no missing values now our model is ready to be used for modelling

```python
# Seperating the target variable from the independent variable
x = data.drop('crime_rate',axis = 1)
y = data['crime_rate']
```

```python
# Splitting the independent variables and dependent variables into test set and training set
from sklearn import model_selection
x_train,x_test,y_train,y_test = model_selection.train_test_split(x,y,test_size = 0.3)
```

```python
# Now let's build our first decision tree
from sklearn.tree import DecisionTreeRegressor
model = DecisionTreeRegressor(random_state = 100)
model.fit(x_train,y_train)
```

```python
# Let's check the model accuracy using the accuracy score
from sklearn.metrics import mean_squared_error
print("The accuracy score for the train data is ",mean_squared_error(y_train,model.predict(x_train)))
```

```python
print("The mean square error for the test data is ",mean_squared_error(y_test,model.predict(x_test)))
```

```python
# Improving the model
model = DecisionTreeRegressor(max_depth = 3)
model.fit(x_train,y_train)
```

```python
# Lets check if the model has now improved
print("The mean squared error for the training model is ",mean_squared_error(y_train,model.predict(x_train)))
print("The mean squared error for the testing model is ",mean_squared_error(y_test,model.predict(x_test)))
```
